# Olist Orders Dataset Inspection

## Objective
Load and inspect the Olist orders CSV before analysis.

## Dataset
Olist Brazilian E-Commerce orders dataset.

## Initial focus
- Preview the first five rows.
- Check the dataset dimensions and columns.

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent
orders_path = project_root / "data" / "raw" / "olist_orders_dataset.csv"

if not orders_path.is_file():
    raise FileNotFoundError(
        f"Olist orders CSV not found: {orders_path}. "
        "Place olist_orders_dataset.csv directly in data/raw/."
    )

orders = pd.read_csv(orders_path)
print(f"Loaded: {orders_path}")
print(f"Shape: {orders.shape}")
print(f"Columns: {orders.columns.tolist()}")
display(orders.head())
```
